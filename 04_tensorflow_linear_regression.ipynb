```python
import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt

# 1. Generación de Datos
# Crearemos una relación simple: y = 2x - 1
X = np.array([-1.0, 0.0, 1.0, 2.0, 3.0, 4.0], dtype=float)
y = np.array([-3.0, -1.0, 1.0, 3.0, 5.0, 7.0], dtype=float)

# 2. Definición del Modelo
# Una sola capa 'Dense' con 1 neurona es una Regresión Lineal
model = tf.keras.Sequential([
    tf.keras.layers.Dense(units=1, input_shape=[1])
])

# 3. Compilación
# 'sgd' = Descenso de Gradiente Estocástico
# 'mean_squared_error' = MSE
model.compile(optimizer='sgd', loss='mean_squared_error')

# 4. Entrenamiento
print("Entrenando modelo...")
# Entrenamos por 500 épocas (iteraciones)
history = model.fit(X, y, epochs=500, verbose=0)
print("Entrenamiento finalizado.")

# 5. Predicción
resultado = model.predict(np.array([10.0]))
print(f"Predicción para x=10: {resultado[0][0]}")
# Debería ser cercano a 19 (y = 2*10 - 1 = 19)

# 6. Visualización de la pérdida (MSE) durante el entrenamiento
plt.xlabel('Época')
plt.ylabel('Pérdida (MSE)')
plt.plot(history.history['loss'])
plt.title('Descenso del Error durante el entrenamiento')
plt.show()
```
