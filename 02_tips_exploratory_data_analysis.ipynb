```python
import seaborn as sns
import matplotlib.pyplot as plt

# Cargar el dataset "tips"
tips = sns.load_dataset("tips")

sns.pairplot(tips, hue="sex") # 'hue' añade color según el sexo
plt.show()
```

```python
# Calculamos la matriz de correlación (solo valores numéricos)
corr = tips.select_dtypes(include=['float64', 'int64']).corr()

sns.heatmap(corr, annot=True, cmap='coolwarm')
plt.title("Correlación entre variables")
plt.show()
```

```python
# Distribución de propinas por día de la semana, diferenciando por sexo
sns.boxplot(data=tips, x="day", y="tip", hue="sex")
plt.title("Distribución de Propinas por Día y Sexo")
plt.show()
```

```python
# Número de clientes (registros) por sexo
sns.countplot(data=tips, x="sex", palette="viridis")
plt.title("Cantidad de Clientes por Sexo")
plt.show()
```
