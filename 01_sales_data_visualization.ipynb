```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('/content/sales_data.csv')
ventas_product = df.groupby('Product_Category')['Sales_Amount'].sum()
ventas_product.plot(kind='bar')
plt.title('Ventas Totales por Región')
plt.xlabel('producto')
plt.ylabel('Ventas')
plt.xticks(rotation=0)
plt.show()
```

```python
ventas_region = df.groupby('Region')['Sales_Amount'].sum()
ventas_region.plot(kind='bar')
plt.title('Ventas Totales por Región')
plt.xlabel('Región')
plt.ylabel('Ventas')
plt.xticks(rotation=0)
plt.show()
```

```python
df_food = df[df['Product_Category'] == 'Food']
ventas_food_region = df_food.groupby('Region')['Sales_Amount'].sum()

ventas_food_region.plot(
    kind='pie',
    autopct='%1.1f%%',
    startangle=90
)

plt.title('Ventas de Comida por Región')
plt.ylabel('')  # Quita el label del eje Y
plt.axis('equal')  # Hace el pastel circular
plt.show()
```

```python
ventas_food_region = ventas_food_region.sort_values(ascending=False)

ventas_food_region.plot(kind='bar')
plt.title('Ventas de Comida por Región')
plt.xlabel('Región')
plt.ylabel('Ventas')
plt.show()
```
