## Try another natural language query


```python
from src.corpus import load_documents
from src.search import SearchIndex
index = SearchIndex(load_documents())
query = 'How can a teacher adapt a reading lesson?'
print('Query:', query)
for match in index.search(query):
    print(f"{match['id']} | {match['score']:.4f} | {match['title']}")
```

## Limits

These are synthetic examples. The reported perfect accuracy does not establish performance on real organizational documents. TF-IDF uses shared words and cannot reliably match arbitrary synonyms. The conceptual graph is a design artifact and is not used as a model feature.
