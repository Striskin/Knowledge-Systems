```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.tag import pos_tag
from nltk.chunk import ne_chunk

# 1. Descargas necesarias (¡Vital para evitar errores!)
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('averaged_perceptron_tagger_eng')
nltk.download('maxent_ne_chunker_tab') # Para NER
nltk.download('words')                 # Diccionario de palabras para NER

# 2. Texto de ejemplo (+50 palabras)
text = """
Alice and Bob traveled to Paris last summer to visit the headquarters of
Google and Microsoft. They met with Steve Jobs' former colleagues to discuss
the future of technology in France. While walking through the Louvre Museum,
they realized that artificial intelligence is changing how people work in
London and New York. The scientists at NASA are also studying these trends
to improve space exploration.
"""

# --- TAREA 1: POS Tagging (Sustantivos y Verbos) ---
tokens = word_tokenize(text)
tags = pos_tag(tokens)

# Filtramos sustantivos (NN) y verbos (VB)
nouns = [word for word, tag in tags if tag.startswith('NN')]
verbs = [word for word, tag in tags if tag.startswith('VB')]

# --- TAREA 2: Named Entity Recognition (NER) ---
# ne_chunk identifica entidades como PERSON, ORGANIZATION y GPE (lugares)
tree = ne_chunk(tags)

print(f"--- RESULTADOS POS TAGGING ---")
print(f"Sustantivos encontrados ({len(nouns)}): {nouns[:5]}...")
print(f"Verbos encontrados ({len(verbs)}): {verbs[:5]}...")

print(f"\n--- RESULTADOS NER ---")
for chunk in tree:
    if hasattr(chunk, 'label'):
        # Imprimimos la entidad y su categoría
        entity = " ".join([c[0] for c in chunk])
        print(f"Entidad: {entity:<15} | Tipo: {chunk.label()}")
```

```python
import nltk
import random
from nltk.corpus import movie_reviews

# 1. Descargar el dataset de críticas de cine
nltk.download('movie_reviews')
nltk.download('punkt')

# 2. Preparar los datos
# Creamos una lista de tuplas: (palabras_del_review, categoría)
documents = [(list(movie_reviews.words(fileid)), category)
             for category in movie_reviews.categories()
             for fileid in movie_reviews.fileids(category)]

# Mezclamos para que el entrenamiento sea efectivo
random.shuffle(documents)

# 3. Definir el extractor de características
# Buscamos las 2000 palabras más comunes para usarlas como referencia
all_words = nltk.FreqDist(w.lower() for w in movie_reviews.words())
word_features = list(all_words)[:2000]

def document_features(document):
    document_words = set(document)
    features = {}
    for word in word_features:
        features['contains({})'.format(word)] = (word in document_words)
    return features

# 4. Crear el conjunto de entrenamiento y prueba
featuresets = [(document_features(d), c) for (d, c) in documents]
train_set, test_set = featuresets[100:], featuresets[:100]

# 5. Entrenar el clasificador (Naive Bayes es el estándar en NLTK)
classifier = nltk.NaiveBayesClassifier.train(train_set)

# 6. Evaluar la precisión (Accuracy)
accuracy = nltk.classify.accuracy(classifier, test_set)

print(f"--- RESULTADOS DEL CLASIFICADOR ---")
print(f"Precisión del modelo: {accuracy * 100:.2f}%")

# Mostrar qué palabras son las más informativas para decidir si es pos/neg
classifier.show_most_informative_features(10)
```

```python
import nltk
import pandas as pd
from nltk.tokenize import word_tokenize

nltk.download('punkt')

# Cargar dataset (limitado a 2000 para velocidad)
df = pd.read_csv('IMDB_dataset-1.csv').sample(2000)

# 1. Tokenización
df['tokens'] = df['review'].apply(lambda x: word_tokenize(x.lower()))

def get_features(words):
    return {word: True for word in words}

featuresets = [(get_features(row['tokens']), row['sentiment']) for index, row in df.iterrows()]
train_set, test_set = featuresets[:1600], featuresets[1600:]
classifier = nltk.NaiveBayesClassifier.train(train_set)

print(f"Accuracy Tokenized: {nltk.classify.accuracy(classifier, test_set)}")
```

```python
import pandas as pd
import re
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Embedding, GlobalAveragePooling1D
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences

# Cargar y preparar datos
df = pd.read_csv('IMDB_dataset-1.csv').sample(5000)

# Funciones de limpieza sin NLTK
def simple_tokenizer(text):
    return re.sub(r'[^\w\s]', '', text.lower()).split()

def simple_lemmatizer(tokens):
    # Reglas básicas (quitar 's' plural, 'ed' pasado, 'ing' gerundio)
    return [re.sub(r'(s|ed|ing)$', '', w) if len(w) > 4 else w for w in tokens]

# Preprocesamiento manual
df['clean_text'] = df['review'].apply(lambda x: " ".join(simple_lemmatizer(simple_tokenizer(x))))

# Preparar para la red neuronal
max_words = 5000
tokenizer = Tokenizer(num_words=max_words)
tokenizer.fit_on_texts(df['clean_text'])
sequences = tokenizer.texts_to_sequences(df['clean_text'])
data = pad_sequences(sequences, maxlen=100)
labels = np.array([1 if s == 'positive' else 0 for s in df['sentiment']])

# Crear el modelo de Red Neuronal
model = Sequential([
    Embedding(max_words, 16, input_length=100),
    GlobalAveragePooling1D(),
    Dense(16, activation='relu'),
    Dense(1, activation='sigmoid')
])

model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
model.fit(data, labels, epochs=5, batch_size=32, verbose=0)

# --- PREDICCIÓN DE 4 COMENTARIOS ---
test_comments = [
    "The movie was absolutely amazing, I loved the acting!",
    "It was a complete waste of time, very boring and long.",
    "Interesting plot but the ending was very disappointing.",
    "A masterpiece of cinema, definitely worth watching twice."
]

# Procesar y predecir
clean_test = [" ".join(simple_lemmatizer(simple_tokenizer(c))) for c in test_comments]
test_seq = pad_sequences(tokenizer.texts_to_sequences(clean_test), maxlen=100)
predictions = model.predict(test_seq)

print("\n--- Resultados de Predicción (Red Neuronal) ---")
for i, comment in enumerate(test_comments):
    sentiment = "Positivo" if predictions[i] > 0.5 else "Negativo"
    print(f"Comentario: {comment}\nResultado: {sentiment} ({predictions[i][0]:.2f})\n")
```
