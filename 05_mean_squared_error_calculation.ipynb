```python
import numpy as np

# 1. Definir los datos de entrada (x) y los valores reales (y)
x = np.array([1, 2, 3])
y_real = np.array([2, 4, 5])

# 2. Parámetros del modelo (pesos y sesgo)
w = 0.5
b = 0.2

# 3. Calcular las predicciones (y_hat) usando la fórmula: y = wx + b
y_pred = w * x + b

# 4. Calcular el Error Cuadrático Medio (MSE)
# Fórmula: MSE = (1/n) * Σ(y_real - y_pred)^2
n = len(y_real)
mse = np.mean((y_real - y_pred)**2)

# --- Resultados ---
print(f"Predicciones (y_hat): {y_pred}")
print(f"Valores Reales (y):    {y_real}")
print(f"MSE final: {mse:.4f}")
```
