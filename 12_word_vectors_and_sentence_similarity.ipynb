```python
import matplotlib.pyplot as plt

# 1. Definimos los datos extraídos de tu imagen
datos = {
    "Gato": (2, 8),
    "Perro": (3, 7),
    "Lobo": (4, 6),
    "Gato montés": (3, 5),
    "Pez": (8, 2),
    "Tiburón": (9, 1),
    "sardina": (7, 1),
    "Avión": (1, 9),
    "Pájaro": (5, 8)
}

# 2. Separamos las etiquetas y las coordenadas (X, Y)
etiquetas = list(datos.keys())
x = [coord[0] for coord in datos.values()]
y = [coord[1] for coord in datos.values()]

# 3. Configuramos el tamaño del gráfico
plt.figure(figsize=(10, 8))

# 4. Creamos los puntos en el gráfico
plt.scatter(x, y, color='blue', marker='o', s=100)

# 5. Añadimos el texto a cada punto
for i, etiqueta in enumerate(etiquetas):
    # xytext=(0, 10) mueve el texto un poco arriba del punto para que no se encimen
    plt.annotate(etiqueta, (x[i], y[i]), textcoords="offset points", xytext=(0,10), ha='center', fontsize=12)

# 6. Agregamos detalles visuales (título, etiquetas de los ejes y cuadrícula)
plt.title("Visualización de Palabras (2D)", fontsize=16)
plt.xlabel("Dimensión X (Eje X)")
plt.ylabel("Dimensión Y (Eje Y)")
plt.grid(True, linestyle='--', alpha=0.6)

# Aseguramos que los ejes empiecen desde 0 para mejor contexto
plt.xlim(0, 10)
plt.ylim(0, 10)

# 7. Mostramos el gráfico
plt.show()
```

```python
from sentence_transformers import SentenceTransformer, util

# 1. Cargamos un modelo pre-entrenado que funcione bien en español
# Este modelo es ligero y soporta múltiples idiomas
modelo = SentenceTransformer('paraphrase-multilingual-MiniLM-L12-v2')

# 2. Definimos las oraciones del ejemplo
oracion1 = "El perro toma agua"
oracion2 = "El gato toma agua"
oracion3 = "La paloma vuela por el aire"

# 3. Obtenemos los "embeddings" de cada oración
embedding1 = modelo.encode(oracion1)
embedding2 = modelo.encode(oracion2)
embedding3 = modelo.encode(oracion3)

# 4. Comparamos los embeddings usando Similitud del Coseno
# Esto nos dará un valor entre -1 y 1 (donde 1 es idéntico)

similitud_perro_gato = util.cos_sim(embedding1, embedding2)
similitud_perro_paloma = util.cos_sim(embedding1, embedding3)
similitud_gato_paloma = util.cos_sim(embedding2, embedding3)

# 5. Mostramos los resultados
print("--- Comparación de Embeddings ---")
print(f"Oraciones similares:")
print(f"'{oracion1}' VS '{oracion2}' -> Similitud: {similitud_perro_gato.item():.4f}\n")

print(f"Oraciones distintas:")
print(f"'{oracion1}' VS '{oracion3}' -> Similitud: {similitud_perro_paloma.item():.4f}")
print(f"'{oracion2}' VS '{oracion3}' -> Similitud: {similitud_gato_paloma.item():.4f}")
```
