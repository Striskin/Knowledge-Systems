```python
import re
import random
from datetime import datetime

class SimpleChatBot:
    def __init__(self, name="Bot"):
        self.bot_name = name
        self.user_name = None

        # Lista de intenciones (intent) con patrones y respuestas
        self.intents = [
            {
                "tag": "saludo",
                "patterns": [
                    r"\b(hola|hey|buenas|buenos dias|buenas tardes|buenas noches)\b"
                ],
                "responses": [
                    "¡Hola! ¿Cómo estás?",
                    "¡Hey! ¿Qué tal?",
                    "¡Buenas! ¿En qué te ayudo?"
                ]
            },
            {
                "tag": "despedida",
                "patterns": [
                    r"\b(adios|bye|hasta luego|nos vemos|chao)\b"
                ],
                "responses": [
                    "¡Hasta luego! 👋",
                    "¡Nos vemos! Cuídate.",
                    "¡Bye! Fue un gusto."
                ],
                "end_conversation": True
            },
            {
                "tag": "ayuda",
                "patterns": [
                    r"\b(ayuda|que puedes hacer|opciones|comandos)\b"
                ],
                "responses": [
                    "Puedo: saludarte, decir la hora, contar un chiste, recordar tu nombre y platicar un poco. Prueba: 'hora', 'chiste', 'me llamo Ana'.",
                    "Opciones: saludo, hora, chiste, recordar nombre. Escribe: 'me llamo <nombre>'."
                ]
            },
            {
                "tag": "hora",
                "patterns": [
                    r"\b(hora|que hora es|dime la hora)\b"
                ],
                "responses": [
                    "Son las {time}.",
                    "Ahora mismo: {time}."
                ]
            },
            {
                "tag": "chiste",
                "patterns": [
                    r"\b(chiste|broma|hazme reir)\b"
                ],
                "responses": [
                    "¿Qué hace una abeja en el gimnasio? ¡Zum-ba! 🐝",
                    "¿Por qué el libro se suicidó? Porque tenía demasiadas páginas en blanco 😅",
                    "¿Qué le dijo un bit a otro? Nos vemos en el bus."
                ]
            },
            {
                "tag": "nombre",
                "patterns": [
                    r"\b(me llamo|mi nombre es|soy)\s+([a-zA-ZáéíóúñÁÉÍÓÚÑ]+)\b"
                ],
                "responses": [
                    "¡Encantado, {user}! ¿En qué te ayudo hoy?",
                    "¡Mucho gusto, {user}! ¿Qué quieres hacer?"
                ]
            },
            {
                "tag": "como_estas",
                "patterns": [
                    r"\b(como estas|que tal estas|todo bien)\b"
                ],
                "responses": [
                    "¡Bien! Gracias por preguntar 😊 ¿Y tú?",
                    "Todo bien por acá. ¿Cómo va tu día?"
                ]
            },
        ]

        self.fallback_responses = [
            "No estoy seguro de entender. ¿Puedes reformularlo?",
            "Interesante 🤔. ¿Me lo dices de otra forma?",
            "Aún soy simple. Prueba 'ayuda' para ver qué entiendo."
        ]

    def _match_intent(self, text):
        text = text.lower().strip()

        for intent in self.intents:
            for pattern in intent["patterns"]:
                m = re.search(pattern, text, re.IGNORECASE)
                if m:
                    return intent, m
        return None, None

    def respond(self, text):
        intent, match = self._match_intent(text)

        if not intent:
            # si sabemos el nombre, personalizamos
            prefix = f"{self.user_name}, " if self.user_name else ""
            return prefix + random.choice(self.fallback_responses), False

        # Manejo especial: "nombre"
        if intent["tag"] == "nombre":
            # El grupo 2 captura el nombre
            self.user_name = match.group(2)
            response = random.choice(intent["responses"]).format(user=self.user_name)
            return response, False

        # Manejo especial: "hora"
        if intent["tag"] == "hora":
            now = datetime.now().strftime("%H:%M:%S")
            response = random.choice(intent["responses"]).format(time=now)
            return response, False

        # Respuesta general
        response = random.choice(intent["responses"])

        # Personalización si ya sabemos nombre
        if self.user_name and "{user}" in response:
            response = response.format(user=self.user_name)

        end = intent.get("end_conversation", False)
        return response, end


def main():
    bot = SimpleChatBot(name="SimpleBot")
    print("SimpleBot: ¡Hola! Escribe 'ayuda' para ver opciones. (Escribe 'adios' para salir)")

    while True:
        user = input("Tú: ").strip()
        if not user:
            continue

        response, end = bot.respond(user)
        print(f"SimpleBot: {response}")

        if end:
            break


if __name__ == "__main__":
    main()
```
